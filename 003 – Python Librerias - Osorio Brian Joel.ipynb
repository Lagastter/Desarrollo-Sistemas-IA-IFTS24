## Ejercicio 2 – Dataset simple con Pandas

```
# 1. Creación del DataFrame con los datos de los estudiantes
data = {
    "estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"],
    "nota": [8, 6, 9, 5, 7]
}

df = pd.DataFrame(data)

# 2. Mostrar el DataFrame completo
print("--- DataFrame de Estudiantes ---")
print(df)

# 3. Cálculos estadísticos sobre la columna 'nota'
promedio = df["nota"].mean()
nota_maxima = df["nota"].max()
nota_minima = df["nota"].min()

# 4. Mostrar resultados por pantalla
print("\n--- Estadísticas de Notas ---")
print(f"Promedio de las notas: {promedio}")
print(f"Nota máxima: {nota_maxima}")
print(f"Nota mínima: {nota_minima}")
```
